```python
def predict_machine_failure(
    machine_type,
    air_temperature,
    process_temperature,
    rotational_speed,
    torque,
    tool_wear
):
    machine_data = pd.DataFrame([{
        "Type": machine_type,
        "Air temperature [K]": air_temperature,
        "Process temperature [K]": process_temperature,
        "Rotational speed [rpm]": rotational_speed,
        "Torque [Nm]": torque,
        "Tool wear [min]": tool_wear
    }])

    failure_probability = tuned_model.predict_proba(machine_data)[0, 1]

    predicted_failure = int(failure_probability >= 0.5)

    if predicted_failure == 1:
        risk_level = "High Risk"
    else:
        risk_level = "Low Risk"

    return {
        "failure_probability": failure_probability,
        "predicted_failure": predicted_failure,
        "risk_level": risk_level
    }
```

```python
def predict_machine_failure(
    machine_type,
    air_temperature,
    process_temperature,
    rotational_speed,
    torque,
    tool_wear
):
    # ---------- Input validation ----------
    if machine_type not in ["L", "M", "H"]:
        raise ValueError("Type must be L, M, or H.")

    if air_temperature <= 0:
        raise ValueError("Air temperature must be greater than 0 K.")

    if process_temperature <= 0:
        raise ValueError("Process temperature must be greater than 0 K.")

    if rotational_speed < 0:
        raise ValueError("Rotational speed cannot be negative.")

    if torque < 0:
        raise ValueError("Torque cannot be negative.")

    if tool_wear < 0:
        raise ValueError("Tool wear cannot be negative.")

    # ---------- Create input ----------
    machine_data = pd.DataFrame([{
        "Type": machine_type,
        "Air temperature [K]": air_temperature,
        "Process temperature [K]": process_temperature,
        "Rotational speed [rpm]": rotational_speed,
        "Torque [Nm]": torque,
        "Tool wear [min]": tool_wear
    }])

    # ---------- Prediction ----------
    failure_probability = tuned_model.predict_proba(machine_data)[0, 1]

    predicted_failure = int(failure_probability >= 0.5)

    if predicted_failure == 1:
        risk_level = "High Risk"
    else:
        risk_level = "Low Risk"

    return {
        "failure_probability": failure_probability,
        "predicted_failure": predicted_failure,
        "risk_level": risk_level
    }
```

```python
result = predict_machine_failure(
    machine_type="L",
    air_temperature=300,
    process_temperature=310,
    rotational_speed=1500,
    torque=50,
    tool_wear=100
)

print(result)
```
